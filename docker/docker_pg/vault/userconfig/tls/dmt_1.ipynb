```python
# Sales Analysis Example

# This notebook loads a small sales dataset, calculates totals by product, and visualizes the results.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "product": ["Laptop", "Phone", "Tablet", "Headphones"],
    "units_sold": [120, 350, 80, 210],
    "unit_price": [1200, 700, 450, 150]
}

df = pd.DataFrame(data)
df["revenue"] = df["units_sold"] * df["unit_price"]
df
```

```python
df.plot(
    kind="bar",
    x="product",
    y="revenue",
    legend=False,
    title="Revenue by Product",
    color="skyblue"
)

plt.ylabel("Revenue (PLN)")
plt.tight_layout()
plt.show()
```
